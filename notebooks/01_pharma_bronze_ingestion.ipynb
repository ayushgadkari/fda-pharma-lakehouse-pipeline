```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import current_timestamp

# Initialize Spark Session
spark = SparkSession.builder.appName("Pharma_Bronze_Ingestion").getOrCreate()

# 1. Create Raw Demographic Data (Simulating DEMO23Q4.txt)
demo_data = [
    ("1001", "10011", "20231015", "54", "YR", "M", "US"),
    ("1002", "10021", "20231102", "62", "YR", "F", "CA"),
    ("1003", "10031", "20231120", "45", "YR", "F", "GB"),
    ("1004", "10041", "20231201", "71", "YR", "M", "US"),
    ("1005", "10051", "20231210", "38", "YR", "F", "DE")
]
demo_schema = ["primaryid", "caseid", "event_dt", "age", "age_cod", "gndr_cod", "reporter_country"]

raw_demo_df = spark.createDataFrame(demo_data, demo_schema)
raw_demo_df.withColumn("ingestion_timestamp", current_timestamp()) \
           .write.format("delta").mode("overwrite").saveAsTable("bronze_demographics")

# 2. Create Raw Drug Data (Simulating DRUG23Q4.txt)
drug_data = [
    ("1001", "ASPIRIN", "PS", "ORAL"),
    ("1002", "IBUPROFEN", "PS", "ORAL"),
    ("1003", "LIPITOR", "PS", "ORAL"),
    ("1004", "METFORMIN", "SS", "ORAL"),
    ("1005", "ASPIRIN", "PS", "ORAL")
]
drug_schema = ["primaryid", "drugname", "role_cod", "route"]

raw_drug_df = spark.createDataFrame(drug_data, drug_schema)
raw_drug_df.withColumn("ingestion_timestamp", current_timestamp()) \
          .write.format("delta").mode("overwrite").saveAsTable("bronze_drug")

# 3. Create Raw Reaction Data (Simulating REAC23Q4.txt)
reac_data = [
    ("1001", "NAUSEA"),
    ("1002", "HEADACHE"),
    ("1003", "FATIGUE"),
    ("1004", "DIZZINESS"),
    ("1005", "RASH")
]
reac_schema = ["primaryid", "pt"]

raw_reac_df = spark.createDataFrame(reac_data, reac_schema)
raw_reac_df.withColumn("ingestion_timestamp", current_timestamp()) \
          .write.format("delta").mode("overwrite").saveAsTable("bronze_reaction")

print("Bronze Layer tables successfully created in Delta Lake format!")
```
